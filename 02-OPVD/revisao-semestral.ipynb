```python
valores = [10, 20, 30, 40, 50]
print('Soma:', sum(valores))
print('Média:', sum(valores)/len(valores))
print('Mínimo:', min(valores))
print('Máximo:', max(valores))
```

## 7. Gráficos básicos
Este bloco revisa gráficos de comparação, tendência e proporção, alinhados ao componente de Visualização de Dados.


```python
import matplotlib.pyplot as plt
categorias = ['Jan', 'Fev', 'Mar', 'Abr']
valores = [120, 150, 140, 180]
plt.figure(figsize=(6,4))
plt.bar(categorias, valores)
plt.title('Vendas por mês')
plt.xlabel('Mês')
plt.ylabel('Valor')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(categorias, valores, marker='o')
plt.title('Tendência mensal')
plt.xlabel('Mês')
plt.ylabel('Valor')
plt.show()
```

```python
partes = ['Aprovados', 'Recuperação', 'Faltosos']
quantidades = [20, 8, 2]
plt.figure(figsize=(6,6))
plt.pie(quantidades, labels=partes, autopct='%1.1f%%')
plt.title('Distribuição da turma')
plt.show()
```
