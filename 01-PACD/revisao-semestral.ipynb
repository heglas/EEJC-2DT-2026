## 7. Introdução ao Pandas
Agora vamos revisar como organizar dados em DataFrames.


```python
import pandas as pd
```

```python
dados = {
    'aluno': ['Ana', 'Bruno', 'Carlos', 'Ana'],
    'turma': ['1EM-CD', '1EM-CD', '1EM-CD', '1EM-CD'],
    'media': [7.5, None, 5.0, 7.5]
}
df = pd.DataFrame(dados)
df
```

```python
print(df.isna().sum())
media_geral = df['media'].mean()
df['media'] = df['media'].fillna(media_geral)
df_sem_duplicados = df.drop_duplicates()
df_sem_duplicados
```

```python
agrupado = df_sem_duplicados.groupby('turma')['media'].agg(['count', 'mean', 'min', 'max'])
agrupado
```

```python
dados_frequencia = {
    'aluno': ['Ana', 'Bruno', 'Carlos'],
    'frequencia': [90, 80, 85]
}
df_freq = pd.DataFrame(dados_frequencia)
df_integrado = pd.merge(df_sem_duplicados, df_freq, on='aluno', how='left')
df_integrado
```

## 8. Visualização básica de dados
Revisão de gráficos simples com Matplotlib e Seaborn.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(6,4))
plt.bar(df_integrado['aluno'], df_integrado['media'])
plt.title('Média dos alunos – recuperação')
plt.xlabel('Aluno')
plt.ylabel('Média')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=df_integrado, x='frequencia', y='media', hue='aluno')
plt.title('Relação entre frequência e média')
plt.show()
```
