## Base Teórica

Na ciência de dados, raramente as informações estão em uma só tabela. Usamos o merge() para unir arquivos diferentes através de uma coluna comum (chave).

- O que é: Funciona como um PROCV (Excel) ou JOIN (SQL).

- A lógica: Une linhas de tabelas distintas baseando-se em um código identificador (ex: id_prod).

**O Tipo: how='left'**

Garante que todos os dados da sua tabela principal (esquerda) permaneçam. Se não houver correspondência na segunda tabela, o Pandas preenche com NaN.

**Exemplos**

**1.** Vendas e Produtos (Adicionar nome ao ID).

```python
vendas = pd.DataFrame({'id': [1, 2], 'qtd': [10, 5]})
prods = pd.DataFrame({'id': [1], 'nome': ['Arroz']})

# O ID 2 permanece, mesmo sem nome cadastrado
print(vendas.merge(prods, on='id', how='left'))
```
**2.** Funcionários e Cargos (Relacionar nomes)

```python
func = pd.DataFrame({'nome': ['Ana', 'Beto']})
cargo = pd.DataFrame({'nome': ['Ana'], 'cargo': ['CEO']})

# Beto não é excluído da lista, seu cargo será NaN
print(func.merge(cargo, on='nome', how='left'))
```

## Atividade do Projeto da Unidade EcoMercado Semana 20: Enriquecendo o Relatório de Faturamento

**Contexto:** Na semana anterior, você gerou um relatório com o faturamento total por id_prod. Porém, o dono do EcoMercado não sabe de cor os códigos dos produtos; ele precisa ver o Nome do Produto e a Categoria para tomar decisões.

**Atividade Prática**

1. Carga de Dados: Importe os arquivos produtos.csv e o arquivo gerado na aula anterior relatorio_faturamento_produtos.csv.

2. Integração (Merge): Realize um merge entre esses dois DataFrames utilizando a coluna id_prod como chave e a direção da busca (how) como left.

3. Filtro de Inteligência: Selecione apenas as colunas nome_produto, categoria e Faturamento_Total.

4. Exportação Final: Salve o novo DataFrame como relatorio_final_ecomercado.csv.

````python
#  Importe os arquivos produtos.csv e o arquivo gerado na aula anterior relatorio_faturamento_produtos.csv.
df_produtos = pd.read_csv('produtos.csv')
df_faturamento = pd.read_csv('relatorio_faturamento_produtos.csv')

#Realize um merge entre esses dois DataFrames utilizando a coluna id_prod como chave e a direção da busca (how) como left.
df_final = pd.merge(df_faturamento, df_produtos, on='id_prod', how='left')

#Selecione apenas as colunas nome_produto, categoria e Faturamento_Total.
df_final = df_final[['nome_produto', 'categoria', 'Faturamento_Total']]

#Exportação Final: Salve o novo DataFrame como relatorio_final_ecomercado.csv.
df_final.to_csv('relatorio_final_ecomercado.csv', index=False)

print('Arquivo criado com sucesso!')
````
