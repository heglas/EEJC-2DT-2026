## Objetivos

- Formular uma pergunta investigativa, definir o público e a mensagem principal.
- Criar narrativas com contexto → descoberta/conflito → resolução.
- Preparar bases confiáveis em planilhas.
- Escolher e justificar gráficos de colunas, barras, linhas, pizza e dispersão.
- Personalizar gráficos com clareza, honestidade e acessibilidade.
- Planejar um painel simples para comunicar conclusões com evidências.

## Checklist

- [ ] Números são reconhecidos como números?
- [ ] Há dados ausentes, duplicados ou escritos de modos diferentes?

## Exercício 3 — Funções de apoio

| Indicador | Fórmula sugerida |
|---|---|
| Total de participantes | `=CONT.VALORES(A2:A11)` |
| Quantidade que entregou projeto | `=CONT.SE(D2:D11;"Sim")` |
| Média das notas válidas | `=MÉDIA(E2:E11)` |
| Média de horas em Python | `=MÉDIASE(B2:B11;"Python";C2:C11)` |
| Total de horas de Planilhas | `=SOMASE(B2:B11;"Planilhas";C2:C11)` |

`=SE(D2="Sim";"Projeto entregue";"Pendente")`

# Parte 4 — Personalização e leitura crítica (semana 19)

## Checklist de clareza, honestidade e acessibilidade

- [ ] O título explica o assunto e contribui para a mensagem?
- [ ] Eixos têm nomes e unidades adequadas?
- [ ] Há contraste suficiente entre cores?
- [ ] Cores não são a única forma de diferenciar categorias?
- [ ] Rótulos e legendas ajudam sem poluir o visual?
- [ ] A escala do eixo não exagera pequenas diferenças?
- [ ] Foram evitados 3D, sombras e efeitos decorativos sem função?
- [ ] A fonte e o período dos dados estão indicados?
